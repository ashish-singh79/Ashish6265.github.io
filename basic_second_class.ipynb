np.array() — Apne data se Array banana

Jab hamare paas already numbers/list ho aur usko NumPy array mein convert karna ho, tab np.array() use karte hain.**bold text**

```python
import numpy as np
```

```python
marks = np.array([10,20,30,40])
print(marks)
```

```python
marks = np.array([[20,40],
                  [70,45]])
print(marks)
print(marks.ndim)
```

np.arange() — Range ke numbers ka Array

np.arange() se ek range ke numbers automatically generate hote hain.

```python
marks = np.arange(1,9)
print(marks)
```

```python
marks = np.arange(1,12,3)
print(marks)
```

np.random.randint() — Random Integer

Iska use random whole numbers generate karne ke liye hota hai.

```python
marks = np.random.randint(1,12)
print(marks)
```

```python
marks = np.random.randint(1,14,5)
print(marks)
```

np.random.uniform() — Random Decimal Numbers

randint() se whole numbers milte hain, jabki uniform() se decimal/random floating-point numbers milte hain.

```python
marks = np.random.uniform(1,11)
print(marks)
```

```python
marks = np.random.uniform(1,13,4)
print(marks)
```

np.linspace() — Equal Distance wale Numbers

linspace() ka use given range mein equally spaced numbers banane ke liye hota hai.

Example:

```python
marks = np.linspace(1,14)
print(marks)
```

```python
marks = np.linspace(1,16,3)
print(marks)
```
